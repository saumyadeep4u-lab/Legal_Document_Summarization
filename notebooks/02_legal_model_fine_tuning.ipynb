```python
%cd /content/drive/MyDrive/BIA_Legal_Project/legal_project_unzipped/legal_summarization_project

import os
from google.colab import files

print("⚙️ Updating validation script to output a CSV...")

validation_script = '''"""
src/validate_10_contracts.py
End-to-end validation script. Extracts 10 real contracts, generates AI summaries,
and validates them against the dataset's gold summaries using ROUGE/BERTScore.
Outputs the results to a CSV file.
"""

import logging
import os
from src.data_loader import CUADDataLoader
from src.summarizer import LegalSummarizer
from src.evaluator import LegalEvaluator
import pandas as pd
import warnings

warnings.filterwarnings("ignore")
logging.basicConfig(level=logging.INFO, format="%(message)s")
logger = logging.getLogger(__name__)

def main():
    logger.info("--- 🚀 INITIATING 10-CONTRACT BATCH VALIDATION ---")

    loader = CUADDataLoader()
    df = loader.extract_contracts()

    df["gold_summary"] = df["clauses"].apply(lambda x: " ".join([clause["text"] for clause in x]))
    df_10 = df[df["gold_summary"].str.strip() != ""].head(10).reset_index(drop=True)

    summarizer = LegalSummarizer()
    evaluator = LegalEvaluator()

    rouge_1_total, bertscore_total = 0, 0
    results_list = []

    for idx, row in df_10.iterrows():
        title = row['contract_title']
        full_text = row['full_text']
        gold_summary = row['gold_summary']

        logger.info(f"\\n📄 Processing Contract {idx + 1}/10: {title}")

        ai_summary = summarizer.summarize_document(full_text)
        metrics = evaluator.evaluate_summaries(reference=gold_summary, generated=ai_summary)

        rouge_1_total += metrics["ROUGE-1 (F1)"]
        bertscore_total += metrics["BERTScore (F1)"]

        logger.info(f"   ↳ ROUGE-1: {metrics['ROUGE-1 (F1)']} | BERTScore: {metrics['BERTScore (F1)']}")

        results_list.append({
            "Contract": title,
            "Gold_Summary_Reference": gold_summary,
            "AI_Generated_Summary": ai_summary,
            "ROUGE-1": metrics["ROUGE-1 (F1)"],
            "BERTScore": metrics["BERTScore (F1)"]
        })

    logger.info("\\n==================================================")
    logger.info("✅ 10-CONTRACT VALIDATION COMPLETE")
    logger.info(f"📈 Average ROUGE-1 Score: {round(rouge_1_total / 10, 4)}")
    logger.info(f"📈 Average BERTScore:     {round(bertscore_total / 10, 4)}")
    logger.info("==================================================")

    # SAVE TO CSV
    os.makedirs("data/processed", exist_ok=True)
    results_df = pd.DataFrame(results_list)
    output_path = "data/processed/10_contract_validation_results.csv"
    results_df.to_csv(output_path, index=False)
    logger.info(f"\\n💾 Saved 10-contract proof to: {output_path}")

if __name__ == "__main__":
    main()
'''

# Write the updated script to your files
with open('src/validate_10_contracts.py', 'w') as f:
    f.write(validation_script)

print("🚀 Running the validation (This will take a couple of minutes to process all 10!)...")
!python -m src.validate_10_contracts

# Trigger the browser download!
if os.path.exists("data/processed/10_contract_validation_results.csv"):
    print("\n🎉 Done! Downloading your CSV proof!")
    files.download("data/processed/10_contract_validation_results.csv")
```

```python
import os
import shutil
from google.colab import files

# 1. Set your main project root directory
project_root = "/content/drive/MyDrive/BIA_Legal_Project/legal_project_unzipped/legal_summarization_project"

if not os.path.exists(project_root):
    raise FileNotFoundError(f"Could not find project root at {project_root}. Please check your path.")

os.chdir(project_root)
print(f"📂 Working in: {os.getcwd()}")

# 2. Define a clean temporary export directory
export_dir = "/content/clean_export/legal_summarization_project"
if os.path.exists(export_dir):
    shutil.rmtree(export_dir)
os.makedirs(export_dir, exist_ok=True)

# 3. Specify only the exact items you want included in your final submission
items_to_include = [
    "src",
    "notebooks",
    "deliverables",
    "tests",
    "data/processed",
    "config.yaml",
    "requirements.txt"
]

print("\n📦 Copying only essential project assets...")
for item in items_to_include:
    src_path = os.path.join(project_root, item)
    dst_path = os.path.join(export_dir, item)

    if os.path.exists(src_path):
        if os.path.isdir(src_path):
            shutil.copytree(src_path, dst_path, dirs_exist_ok=True)
            print(f"  ✔ Included folder: {item}/")
        else:
            os.makedirs(os.path.dirname(dst_path), exist_ok=True)
            shutil.copy2(src_path, dst_path)
            print(f"  ✔ Included file: {item}")
    else:
        print(f"  ⚠️ Notice: '{item}' was not found on disk, skipping.")

# 4. Package the clean folder into a zip archive
zip_filename = "/content/legal_summarization_project_SUBMISSION"
print("\n🗜️ Compressing clean project structure...")
shutil.make_archive(zip_filename, 'zip', root_dir='/content/clean_export', base_dir='legal_summarization_project')

# 5. Trigger automatic browser download
print("📥 Downloading your clean submission zip file...")
files.download(f"{zip_filename}.zip")
print("\n✅ Success! Your clean, clutter-free zip file is downloading now.")
```
